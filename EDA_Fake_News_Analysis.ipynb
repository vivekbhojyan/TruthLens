```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Setup styling
sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
print('Environment and libraries ready!')
```

## 1. Load Dataset & Data Dictionary

```python
df = pd.read_csv('dataset_cleaned.csv')
dict_df = pd.read_csv('data_dictionary.csv')

print(f'Total Articles: {len(df)}')
print(f'Columns: {list(df.columns)}')
df.head()
```

```python
# Display data dictionary
dict_df[['column_name', 'data_type', 'nullable', 'description_fill_in']]
```

## 2. Target Variable Exploration (REAL vs FAKE)

```python
palette = {'REAL': '#2ecc71', 'FAKE': '#e74c3c'}
plt.figure(figsize=(7, 5))
ax = sns.countplot(data=df, x='label', palette=palette, order=['REAL', 'FAKE'])
plt.title('Class Distribution: Verified REAL vs Fabricated FAKE', fontsize=13, fontweight='bold')
plt.xlabel('Ground Truth Label')
plt.ylabel('Count')
plt.show()
```

## 3. Numerical Feature Correlations & Statistical Distributions

```python
num_cols = ['word_count', 'sentiment_score', 'exclamation_count', 'all_caps_pct']
corr = df[num_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.3f', square=True)
plt.title('Correlation Heatmap among Article Metadata Features', fontsize=13, fontweight='bold')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=df, x='label', y='exclamation_count', palette=palette, ax=axes[0])
axes[0].set_title('Exclamation Mark Frequency by Article Authenticity')

sns.boxplot(data=df, x='label', y='all_caps_pct', palette=palette, ax=axes[1])
axes[1].set_title('Capitalization Ratio (Clickbait Marker) by Article Authenticity')
plt.tight_layout()
plt.show()
```
